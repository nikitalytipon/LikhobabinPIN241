**Цель работы:** познакомиться с использованием библиотеки pandas при решении практических задач, связанных с анализом данных.

Проанализируем набор данных `Employee.csv`. Датасет содержит сведения о сотрудниках компании. Набор данных включает следующие атрибуты:

- **Education** – уровень образования (Bachelors, Masters, PHD);
- **JoiningYear** – год трудоустройства;
- **City** – город (Bangalore, Pune, New Delhi);
- **PaymentTier** – уровень оплаты (1, 2, 3);
- **Age** – возраст (лет);
- **Gender** – пол (Male, Female);
- **EverBenched** – находился ли в резерве (No/Yes);
- **ExperienceInCurrentDomain** – опыт в текущей области (лет);
- **LeaveOrNot** – уволился или нет (0 – нет, 1 – да).

## 0. Импортируем необходимые библиотеки

```python
import numpy as np
import pandas as pd
```

## 1. Загрузите датасет из файла Employee.csv и поместите его в датафрейм

```python
df = pd.read_csv('Employee.csv')
```

## 8.Проанализируем влияние уровня образования на увольнение.

```python
df.groupby('Education')['LeaveOrNot'].mean()
```

## 9.Создадим возрастную группу и определим, какая возрастная группа увольняется чаще.

```python
df['AgeGroup'] = pd.cut(df['Age'],
                       bins=[0, 30, 45, 100],
                       labels=['<30', '30-45', '45+'])

df.groupby('AgeGroup', observed=False)['LeaveOrNot'].mean()
```

## 10.Проанализируем влияние пола на увольнение и уровень оплаты.

```python
df.groupby('Gender').agg({
    'LeaveOrNot': 'mean',
    'PaymentTier': ['mean', 'median']
})
```

## 11.Оценим влияние уровня оплаты (PaymentTier) на текучесть кадров.

```python
df.groupby('PaymentTier')['LeaveOrNot'].mean()
```

## 12.Рассчитаем общий стаж (Experience) и сгруппируем по стажу, чтобы выявить зависимость между стажем и увольнением.

```python
df['Experience'] = 2026 - df['JoiningYear']

df['ExpGroup'] = pd.cut(df['Experience'],
                       bins=[0, 2, 5, 40],
                       labels=['<2', '2-5', '5+'])

df.groupby('ExpGroup', observed=False)['LeaveOrNot'].mean()
```

## 13.Изучим влияние нахождения на бенче (EverBenched) на увольнение и средний опыт в текущей области.

```python
df.groupby('EverBenched').agg({
    'LeaveOrNot': 'mean',
    'ExperienceInCurrentDomain': 'mean'
})
```

## 14.Построим сводную таблицу: средний возраст и количество сотрудников в зависимости от опыта в текущей области и нахождения на бенче.

```python
pd.pivot_table(df,
               values='Age',
               index='ExperienceInCurrentDomain',
               columns='EverBenched',
               aggfunc=['mean', 'count'])
```

## 15.Построим сводную таблицу: вероятность увольнения в зависимости от пола и уровня образования.

```python
pd.pivot_table(df,
               values='LeaveOrNot',
               index=['Gender', 'Education'],
               aggfunc='mean')
```

## 16.Создадим новый признак — отношение опыта в текущей области к возрасту (experience_to_age_ratio).

```python
df['experience_to_age_ratio'] = df['ExperienceInCurrentDomain'] / df['Age']
df.head()
```

## 17.Сохраним обогащённый датафрейм в CSV-файл для дальнейшего использования.

```python
df.to_csv('result.csv', index=False)
```
