Вывод: определено количество уникальных городов и частота их встречаемости.

```python
df.groupby('Education')['LeaveOrNot'].mean()
```

Вывод: видно, как уровень образования влияет на вероятность увольнения. 
Чем выше значение, тем чаще сотрудники уходят.

```python
df['AgeGroup'] = pd.cut(df['Age'],
                       bins=[0, 30, 45, 100],
                       labels=['<30', '30-45', '45+'])

df.groupby('AgeGroup', observed=False)['LeaveOrNot'].mean()
```

Вывод: определено, какая возрастная группа увольняется чаще.

```python
df.groupby('Gender').agg({
    'LeaveOrNot': 'mean',
    'PaymentTier': ['mean', 'median']
})
```

Вывод: проанализировано влияние пола на увольнение и уровень оплаты.

```python
df.groupby('PaymentTier')['LeaveOrNot'].mean()
```

Вывод: уровень оплаты влияет на текучесть кадров.

```python
df['Experience'] = 2026 - df['JoiningYear']

df['ExpGroup'] = pd.cut(df['Experience'],
                       bins=[0, 2, 5, 40],
                       labels=['<2', '2-5', '5+'])

df.groupby('ExpGroup', observed=False)['LeaveOrNot'].mean()
```

Вывод: выявлена зависимость между стажем и увольнением.

```python
df.groupby('EverBenched').agg({
    'LeaveOrNot': 'mean',
    'ExperienceInCurrentDomain': 'mean'
})
```

Вывод: опыт безработицы влияет на текучесть сотрудников.

```python
pd.pivot_table(df,
               values='Age',
               index='ExperienceInCurrentDomain',
               columns='EverBenched',
               aggfunc=['mean', 'count'])
```

```python
pd.pivot_table(df,
               values='LeaveOrNot',
               index=['Gender', 'Education'],
               aggfunc='mean')
```

Сводные таблицы позволяют одновременно учитывать несколько факторов.

```python
df['experience_to_age_ratio'] = df['ExperienceInCurrentDomain'] / df['Age']
df.head()
```

Создан новый показатель, отражающий отношение опыта к возрасту.

```python
df.to_csv('result.csv', index=False)
```
